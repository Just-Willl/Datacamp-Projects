Every year, American high school students take SATs, which are standardized tests intended to measure literacy, numeracy, and writing skills. There are three sections - reading, math, and writing, each with a **maximum score of 800 points**. These tests are extremely important for students and colleges, as they play a pivotal role in the admissions process.

Analyzing the performance of schools is important for a variety of stakeholders, including policy and education professionals, researchers, government, and even parents considering which school their children should attend. 

You have been provided with a dataset called `schools.csv`, which is previewed below.

You have been tasked with answering three key questions about New York City (NYC) public school SAT performance.

```python
# Re-run this cell 
import pandas as pd

# Read in the data
schools = pd.read_csv("schools.csv")

# Preview the data
schools.head()

# Start coding here...
# Add as many cells as you like...
```

Task 1: Which NYC schools have the best math results? 

```python
# Filter for schools with average_math at least 80% of 800, subset for columns school_name and average_math, sort by average_math decending.
best_math_schools = schools[schools['average_math'] >= 800*0.8][["school_name","average_math",]].sort_values('average_math', ascending=False)
print(best_math_schools)
```

Task 2: What are the top 10 performing schools based on the combined SAT scores? 

```python
# Creating a new column total_SAT with the combined sat scores
schools['total_SAT'] = schools[['average_math','average_reading','average_writing']].sum(axis=1)

# Creating a new DataFrame called top_10_schools with columns school_name and total_SAT sorted by total_SAT decending and only displaying the top 10 results
top_10_schools = schools[["school_name","total_SAT"]].sort_values("total_SAT", ascending=False).head(10)
top_10_schools
```

Task 3: Which single borough has the largest standard deviation in the combined SAT score?

```python
largest_std_dev = schools.groupby('borough')['total_SAT'].agg(['count', 'mean', 'std']).round(2).rename(columns={'count':'num_schools', 'mean':'average_SAT', 'std':'std_SAT'}).sort_values('std_SAT', ascending=False).head(1)
print(largest_std_dev)
```
